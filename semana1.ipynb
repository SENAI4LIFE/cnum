Serão passados os coeficientes a, b e c. Caso a = 0. resolva como uma equação do primeiro grau, e se b = 0 gere uma exceção.


## Importando uma biblioteca

```python
from math import sqrt
```

## Criando uma função
Raiz de equação quadrática

```python
def zeros(a,b,c):

    if a == 0:
         if b == 0:
              raise ValueError("A equação não é de segundo grau")
         else:
              return[-c/b]
        
              




              
    delta = b**2 - (4*a*c)
    if delta < 0:
            return None
    elif delta == 0:
        x = -b / (2*a)
        return[x,x]
    else:
        x1 =(-b +sqrt(delta)) / (2*a)
        x2 =(-b - delta**0.5) / (2*a)
        return[x1,x2]
```
